```python
import os

# Define the base path for the cloned GitHub repository's data directory
# This path is based on where the repository is cloned in Colab.
repo_data_base_path = '/content/UMBC-DATA606-Capstone/data'

# Define the paths for the original input text files (now from the cloned repo)
original_tr_txt_path = os.path.join(repo_data_base_path, 'tr_txt_table.txt')
original_va_txt_table_path = os.path.join(repo_data_base_path, 'va_txt_table.txt')
original_te_txt_table_path = os.path.join(repo_data_base_path, 'te_txt_table.txt')

# Define the paths for the new tokenized output files (now to the cloned repo)
formatted_tr_txt_path = os.path.join(repo_data_base_path, 'train.txt')
formatted_va_txt_path = os.path.join(repo_data_base_path, 'val.txt')
formatted_te_txt_path = os.path.join(repo_data_base_path, 'test.txt')

# Image IDs corresponding to the files (these are used in the output file names)
train_image_id = 'train'
validate_image_id = 'validate'
test_image_id = 'test'

# Function to read content and write in the required format (tokenizing into space-separated characters)
def format_text_table(original_path, formatted_path, image_id):
    # Debugging: Check if the file exists before trying to open it
    if not os.path.exists(original_path):
        print(f"DEBUG: File does not exist according to os.path.exists(): {original_path}")
        # Re-raise or handle as before, but this message helps confirm the root cause

    try:
        with open(original_path, 'r', encoding='utf-8') as f_orig:
            raw_content = f_orig.read().strip()

            token_list = []
            if not raw_content:
                tokenized_transcription = ""
            else:
                for char in raw_content:
                    if char.isspace(): # Treat any whitespace as a single <space> token
                        # Avoid adding multiple <space> tokens if there are consecutive spaces
                        if not token_list or token_list[-1] != '<space>':
                            token_list.append('<space>')
                    else:
                        token_list.append(char)

                # Remove leading/trailing <space> if any, as strip() was called on raw_content.
                # Only remove if the list is not empty after tokenization.
                if token_list and token_list[0] == '<space>':
                    token_list.pop(0)
                if token_list and token_list[-1] == '<space>':
                    token_list.pop()

                tokenized_transcription = ' '.join(token_list)

        if not tokenized_transcription.strip(): # Check if transcription is effectively empty
            print(f"Warning: The original file '{original_path}' resulted in an empty or whitespace-only transcription for image ID '{image_id}'. This sample might be ignored during training.")

        # Write the formatted content to the new file
        with open(formatted_path, 'w', encoding='utf-8') as f_formatted:
            f_formatted.write(f"{image_id} {tokenized_transcription}\n") # Add newline at the end of the entry
        print(f"Successfully created formatted text table: {formatted_path}")
    except FileNotFoundError:
        print(f"Error: Original file not found at {original_path}. Please ensure this file exists in your cloned GitHub repository's data directory.")
    except Exception as e:
        print(f"An error occurred while formatting {original_path}: {e}")

# Format the training text table
format_text_table(original_tr_txt_path, formatted_tr_txt_path, train_image_id)

# Format the validation text table
format_text_table(original_va_txt_table_path, formatted_va_txt_path, validate_image_id)

# Format the test text table
format_text_table(original_te_txt_table_path, formatted_te_txt_path, test_image_id)

# Update the global variables in the current session to point to the new tokenized files
global tr_txt_table_path
global va_txt_table_path
global te_txt_table_path
tr_txt_table_path = formatted_tr_txt_path
va_txt_table_path = formatted_va_txt_path
te_txt_table_path = formatted_te_txt_path

print(f"Updated tr_txt_table_path to: {tr_txt_table_path}")
print(f"Updated va_txt_table_path to: {va_txt_table_path}")
print(f"Updated te_txt_table_path to: {te_txt_table_path}")
```

```python
import os
from PIL import Image

# Define paths
original_img_dir = '/content/UMBC-DATA606-Capstone/data'
resized_img_dir = '/content/resized_images'
input_image_name = 'test.jpeg'

# Create the directory for resized images if it doesn't exist
os.makedirs(resized_img_dir, exist_ok=True)

# Full path to the original image
original_img_path = os.path.join(original_img_dir, input_image_name)
resized_img_path = os.path.join(resized_img_dir, input_image_name)

# Resize the image
if os.path.exists(original_img_path):
    with Image.open(original_img_path) as img:
        # Explicitly convert to grayscale 'L' mode, as HTR models typically expect this
        img = img.convert('L')

        original_width, original_height = img.size
        # Target height should be 128 pixels so that after 3 pooling layers (factor of 8),
        # the feature map height becomes 16, which the sequencer expects.
        target_height = 128
        # Calculate new width to maintain aspect ratio
        new_width = int(original_width * (target_height / original_height))
        if new_width == 0: # Ensure width is at least 1 pixel
            new_width = 1

        resized_img = img.resize((new_width, target_height), Image.LANCZOS)

        # Save the resized image to the new directory
        resized_img.save(resized_img_path)
        print(f"Resized image saved to: {resized_img_path}")

        # VERIFICATION STEP: Read the saved image and check its dimensions
        with Image.open(resized_img_path) as check_img:
            print(f"Verification: Saved image dimensions: {check_img.size}")

else:
    print(f"Error: Original image not found at {original_img_path}")

# Update img_list.txt to use the full filename with extension
img_list_content = input_image_name # Should be 'test.jpeg'
with open('/content/img_list.txt', 'w') as f:
    f.write(img_list_content + '\n')
print(f"Updated img_list.txt with: {img_list_content}")

# Create a new config file for the Hugging Face model
# Now pointing img_dirs to the directory containing the resized image
# Corrected paths to the Hugging Face model files
config_huginmunin_content = f"""
syms: /content/UMBC-DATA606-Capstone/pylaia-huginmunin/syms.txt
img_list: img_list.txt
img_dirs:
  - {resized_img_dir}
common:
  checkpoint: /content/UMBC-DATA606-Capstone/pylaia-huginmunin/weights.ckpt
  model_filename: /content/UMBC-DATA606-Capstone/pylaia-huginmunin/model
decode:
  join_string: ""
  convert_spaces: true
trainer:
  gpus: 0
"""

with open('config_decode_huginmunin.yaml', 'w') as f:
    f.write(config_huginmunin_content)

# Run pylaia-htr-decode-ctc with the new config file
! pylaia-htr-decode-ctc --config config_decode_huginmunin.yaml
```
